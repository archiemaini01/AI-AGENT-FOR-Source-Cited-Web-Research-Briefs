````python
# ============================================================
# CELL 1 — SETUP + RESEARCH TOOLS
# ============================================================

!pip -q install -U "pydantic-ai-slim[openai]" duckduckgo-search requests beautifulsoup4

import os
import json
import re
import pathlib
from typing import Any, Dict, List
from urllib.parse import urlparse, quote

import requests
from bs4 import BeautifulSoup

# ============================================================
# OPENROUTER API KEY
# ============================================================

from getpass import getpass

if not os.environ.get("OPENROUTER_API_KEY"):
    os.environ["OPENROUTER_API_KEY"] = getpass(
        "Enter your OpenRouter API key: "
    )

OPENROUTER_BASE_URL = "https://openrouter.ai/api/v1"
OPENROUTER_MODEL = "openai/gpt-4o-mini"


# ============================================================
# PYDANTICAI + OPENROUTER
# ============================================================

from pydantic_ai import Agent
from pydantic_ai.providers.openai import OpenAIProvider
from pydantic_ai.models.openai import OpenAIChatModel

# Current PydanticAI API:
# OpenRouter is OpenAI-compatible, so use OpenAIProvider
openrouter_provider = OpenAIProvider(
    base_url=OPENROUTER_BASE_URL,
    api_key=os.environ["OPENROUTER_API_KEY"],
)

OPENROUTER_MODEL_INSTANCE = OpenAIChatModel(
    OPENROUTER_MODEL,
    provider=openrouter_provider,
)

print("PydanticAI/OpenRouter model initialized successfully.")


# ============================================================
# ARTIFACT DIRECTORY
# ============================================================

DATA_DIR = pathlib.Path(".").resolve()
ARTIFACTS_DIR = DATA_DIR / "artifacts"
ARTIFACTS_DIR.mkdir(
    exist_ok=True
)


# ============================================================
# DUCKDUCKGO
# ============================================================

try:
    from duckduckgo_search import DDGS
except Exception:
    DDGS = None


# ============================================================
# Q1 — DUCKDUCKGO SEARCH
# ============================================================

def ddg_search(
    query: str,
    max_results: int = 10
) -> List[Dict[str, Any]]:

    results = []

    if DDGS is None:
        return results

    try:
        with DDGS() as ddgs:

            raw_results = ddgs.text(
                query,
                max_results=max_results
            )

            for item in raw_results:

                results.append({
                    "title": item.get(
                        "title",
                        ""
                    ),
                    "url": item.get(
                        "href",
                        item.get("url", "")
                    ),
                    "snippet": item.get(
                        "body",
                        item.get("snippet", "")
                    ),
                })

    except Exception as e:

        print(
            "DuckDuckGo search error:",
            e
        )

    return results


# ============================================================
# Q1 — WIKIPEDIA
# ============================================================

def wiki_summary(
    title_or_query: str,
    sentences: int = 3
) -> Dict[str, str]:

    try:

        title = title_or_query.replace(
            " ",
            "_"
        )

        url = (
            "https://en.wikipedia.org/api/rest_v1/page/summary/"
            + quote(
                title,
                safe="_()"
            )
        )

        response = requests.get(
            url,
            timeout=15,
            headers={
                "User-Agent": "Mozilla/5.0"
            }
        )

        # ----------------------------------------------------
        # If direct page lookup fails, search Wikipedia
        # ----------------------------------------------------

        if response.status_code != 200:

            search_response = requests.get(
                "https://en.wikipedia.org/w/api.php",
                params={
                    "action": "query",
                    "list": "search",
                    "srsearch": title_or_query,
                    "format": "json",
                    "utf8": 1,
                },
                timeout=15,
                headers={
                    "User-Agent": "Mozilla/5.0"
                },
            )

            data = search_response.json()

            search_items = data.get(
                "query",
                {}
            ).get(
                "search",
                []
            )

            if not search_items:

                return {
                    "title": title_or_query,
                    "url": "",
                    "summary": ""
                }

            title = search_items[0][
                "title"
            ].replace(
                " ",
                "_"
            )

            url = (
                "https://en.wikipedia.org/api/rest_v1/page/summary/"
                + quote(
                    title,
                    safe="_()"
                )
            )

            response = requests.get(
                url,
                timeout=15,
                headers={
                    "User-Agent": "Mozilla/5.0"
                }
            )

        data = response.json()

        full_summary = data.get(
            "extract",
            ""
        )

        sentences_list = re.split(
            r'(?<=[.!?])\s+',
            full_summary.strip()
        )

        summary = " ".join(
            sentences_list[
                :max(1, sentences)
            ]
        )

        return {
            "title": data.get(
                "title",
                title_or_query
            ),
            "url": data.get(
                "content_urls",
                {}
            ).get(
                "desktop",
                {}
            ).get(
                "page",
                f"https://en.wikipedia.org/wiki/{title}"
            ),
            "summary": summary,
        }

    except Exception as e:

        return {
            "title": title_or_query,
            "url": "",
            "summary": (
                f"Unable to retrieve "
                f"Wikipedia summary: {e}"
            )
        }


# ============================================================
# Q1 — FETCH AND CLEAN WEB PAGE
# ============================================================

def fetch_page_clean(
    url: str
) -> Dict[str, str]:

    try:

        response = requests.get(
            url,
            timeout=20,
            headers={
                "User-Agent": (
                    "Mozilla/5.0 "
                    "(Windows NT 10.0; Win64; x64) "
                    "AppleWebKit/537.36 "
                    "Chrome/130 Safari/537.36"
                )
            }
        )

        response.raise_for_status()

        soup = BeautifulSoup(
            response.text,
            "html.parser"
        )

        # Remove unnecessary HTML
        for tag in soup([
            "script",
            "style",
            "noscript",
            "nav",
            "footer",
            "header",
            "aside",
            "form"
        ]):
            tag.decompose()

        title = (
            soup.title.get_text(
                strip=True
            )
            if soup.title
            else url
        )

        text = soup.get_text(
            separator=" ",
            strip=True
        )

        text = re.sub(
            r"\s+",
            " ",
            text
        ).strip()

        # Keep response manageable
        text = text[:12000]

        return {
            "title": title,
            "url": url,
            "text": text,
        }

    except Exception as e:

        return {
            "title": url,
            "url": url,
            "text": (
                f"Unable to fetch page: {e}"
            ),
        }


# ============================================================
# Q1 — SELECT TOP URLS
# ============================================================

def select_top_urls(
    results: List[Dict[str, Any]],
    k: int = 5
) -> List[str]:

    urls = []
    seen = set()

    for item in results:

        url = item.get(
            "url"
        ) or item.get(
            "href"
        )

        if not url:
            continue

        if url in seen:
            continue

        if not url.startswith(
            ("http://", "https://")
        ):
            continue

        seen.add(url)
        urls.append(url)

        if len(urls) >= k:
            break

    return urls


# ============================================================
# Q1 — SAVE LOGS
# ============================================================

def save_logs(
    payload: Dict[str, Any],
    path: str
) -> str:

    output_path = pathlib.Path(path)

    if (
        not output_path.is_absolute()
        and output_path.parent == pathlib.Path(".")
    ):
        output_path = (
            ARTIFACTS_DIR /
            output_path
        )

    output_path.parent.mkdir(
        parents=True,
        exist_ok=True
    )

    with open(
        output_path,
        "w",
        encoding="utf-8"
    ) as f:

        json.dump(
            payload,
            f,
            indent=2,
            ensure_ascii=False,
            default=str
        )

    return str(output_path)


print()
print("=" * 60)
print("CELL 1 COMPLETE")
print("=" * 60)
print("OpenRouter model:", OPENROUTER_MODEL)
print("Artifacts:", ARTIFACTS_DIR)
print("All Q1 functions loaded.")
````

````python
# ============================================================
# CELL 2 — PYDANTICAI AGENT + REACT + SYNTHESIS
# ============================================================

SYSTEM_INSTRUCTIONS = """
You are a conversational ReAct-style research agent.

Your job is to research a user's topic using the available tools
and then create a reliable source-cited research brief.

TOOL USAGE:
1. Use ddg_search for broad web research and discovery.
2. Use wiki_summary when Wikipedia is useful for background/context.
3. Use fetch_page_clean to inspect important web pages in detail.
4. Use select_top_urls to select the strongest sources.
5. Do not invent sources or URLs.

RESEARCH REQUIREMENTS:
- Prefer reliable and diverse sources.
- Try to obtain at least 3 different domains when possible.
- Use multiple sources to verify important claims.
- Keep track of titles, URLs, and useful snippets.
- Use source material rather than unsupported assumptions.

FINAL BRIEF REQUIREMENTS:
- Write 5–8 clear paragraphs.
- Include inline numeric citations such as [1], [2], [3].
- Include at least two short direct quotes.
- Every direct quote must be 10 words or fewer.
- Add a References section.
- References must map citation numbers to source titles and URLs.
- Never fabricate quotations.
- Never fabricate URLs.
- Keep the writing concise and factual.

When you have enough evidence, synthesize the research.
"""


# ------------------------------------------------------------
# Agent
# ------------------------------------------------------------

agent = Agent(
    OPENROUTER_MODEL_INSTANCE,
    system_prompt=SYSTEM_INSTRUCTIONS,
)


# ------------------------------------------------------------
# Tool wrappers
# ------------------------------------------------------------

@agent.tool_plain
def ddg_search_tool(
    query: str,
    max_results: int = 10
) -> List[Dict[str, Any]]:
    """Search DuckDuckGo."""
    return ddg_search(
        query,
        max_results=max_results
    )


@agent.tool_plain
def wiki_summary_tool(
    title_or_query: str,
    sentences: int = 3
) -> Dict[str, str]:
    """Get a short Wikipedia summary."""
    return wiki_summary(
        title_or_query,
        sentences=sentences
    )


@agent.tool_plain
def fetch_page_clean_tool(
    url: str
) -> Dict[str, str]:
    """Fetch and clean a web page."""
    return fetch_page_clean(url)


@agent.tool_plain
def select_top_urls_tool(
    results: List[Dict[str, Any]],
    k: int = 5
) -> List[str]:
    """Select the strongest URLs from search results."""
    return select_top_urls(
        results,
        k=k
    )


@agent.tool_plain
def save_logs_tool(
    payload: Dict[str, Any],
    path: str
) -> str:
    """Save research logs."""
    return save_logs(
        payload,
        path
    )


# ============================================================
# Q3 — RUN AGENT
# ============================================================

def run_agent(
    topic: str,
    rounds: int = 1
) -> Dict[str, Any]:

    search_results = []
    selected_urls = []
    snippets = []
    messages = []
    draft = ""

    message_history = None

    for round_number in range(max(1, rounds)):

        if round_number == 0:
            prompt = f"""
Research the following topic:

{topic}

Perform actual web research before writing the draft.

Requirements:
- Use DuckDuckGo.
- Use Wikipedia when useful.
- Inspect several useful pages.
- Try to use at least 3 different domains.
- Select the strongest URLs.
- Gather useful snippets/evidence.
- Then write a preliminary research draft.
"""
        else:
            prompt = f"""
Continue researching and improving the research brief about:

{topic}

Use the previous conversation as context.

Check whether:
- enough independent sources were used,
- at least 3 domains are represented,
- important claims have citations,
- at least two short quotations are available,
- quotations are no more than 10 words,
- references can be mapped to URLs.

Then produce an improved draft.
"""

        try:
            if message_history is None:
                result = agent.run_sync(prompt)
            else:
                result = agent.run_sync(
                    prompt,
                    message_history=message_history
                )

            draft = getattr(
                result,
                "output",
                str(result)
            )

            # Required by the assignment:
            # multi-turn PydanticAI message history
            try:
                message_history = result.new_messages()
            except Exception:
                message_history = None

            try:
                messages.extend(
                    [str(m) for m in result.new_messages()]
                )
            except Exception:
                pass

        except Exception as e:
            draft = f"Research agent error: {e}"

    # --------------------------------------------------------
    # Independently collect source information as a safety net.
    # This also ensures the report contains useful logs.
    # --------------------------------------------------------

    try:
        search_results = ddg_search(
            topic,
            max_results=10
        )
    except Exception:
        search_results = []

    try:
        selected_urls = select_top_urls(
            search_results,
            k=5
        )
    except Exception:
        selected_urls = []

    # Fetch selected sources
    source_documents = []

    for url in selected_urls:
        try:
            doc = fetch_page_clean(url)

            source_documents.append(doc)

            text = doc.get("text", "")

            snippets.append({
                "title": doc.get("title", ""),
                "url": url,
                "snippet": text[:1000]
            })

        except Exception:
            pass

    # Add Wikipedia background
    try:
        wiki = wiki_summary(topic, sentences=3)

        if wiki.get("url"):
            snippets.append({
                "title": wiki.get("title", ""),
                "url": wiki.get("url", ""),
                "snippet": wiki.get("summary", "")
            })

    except Exception:
        pass

    report = {
        "topic": topic,
        "messages": messages,
        "search_results": search_results,
        "selected_urls": selected_urls,
        "snippets": snippets,
        "draft": draft,
    }

    # Save required report
    save_logs(
        report,
        str(ARTIFACTS_DIR / "report.json")
    )

    return report


# ============================================================
# Q4 — SYNTHESIZE BRIEF
# ============================================================

def _count_words(text: str) -> int:
    return len(re.findall(r"\b\w+\b", text))


def _extract_sources(report: Dict[str, Any]) -> List[Dict[str, str]]:
    """
    Build a clean numbered source list.
    """
    sources = []
    seen = set()

    # Prefer snippets because they often contain titles + URLs
    for item in report.get("snippets", []):
        url = item.get("url", "")

        if not url or url in seen:
            continue

        seen.add(url)

        sources.append({
            "title": item.get("title", url),
            "url": url,
        })

    # Add selected URLs not already represented
    for url in report.get("selected_urls", []):
        if not url or url in seen:
            continue

        seen.add(url)

        sources.append({
            "title": urlparse(url).netloc,
            "url": url,
        })

    return sources


def synthesize_brief(
    report: Dict[str, Any],
    min_paras: int = 5,
    max_paras: int = 8
) -> Dict[str, Any]:

    topic = report.get("topic", "Research Topic")
    draft = report.get("draft", "")
    sources = _extract_sources(report)

    # --------------------------------------------------------
    # Ask the model to create the final brief
    # --------------------------------------------------------

    source_text = "\n".join(
        f"[{i+1}] {s['title']} — {s['url']}"
        for i, s in enumerate(sources)
    )

    snippets_text = "\n\n".join(
        f"Source: {x.get('title', '')}\n"
        f"URL: {x.get('url', '')}\n"
        f"Evidence: {x.get('snippet', '')[:1200]}"
        for x in report.get("snippets", [])
    )

    synthesis_prompt = f"""
Create the final research brief for:

{topic}

Use ONLY the research evidence below.

DRAFT:
{draft}

SOURCES:
{source_text}

EVIDENCE:
{snippets_text}

STRICT FORMAT:
- Write exactly 5–8 substantive paragraphs before References.
- Use inline citations [1], [2], [3], etc.
- Use at least 3 different domains when the supplied sources allow it.
- Include at least TWO short direct quotations.
- Each quotation must be 10 words or fewer.
- Do not invent quotations.
- Do not invent facts.
- End with:

References

[1] Title — URL
[2] Title — URL

Continue numbering for every source actually cited.

Do not put the References heading inside a paragraph.
"""

    try:
        synthesis_agent = Agent(
            OPENROUTER_MODEL_INSTANCE,
            system_prompt=SYSTEM_INSTRUCTIONS
        )

        result = synthesis_agent.run_sync(
            synthesis_prompt
        )

        brief = getattr(
            result,
            "output",
            str(result)
        )

    except Exception:
        brief = draft

    # --------------------------------------------------------
    # Clean accidental markdown fences
    # --------------------------------------------------------

    brief = re.sub(
        r"^```(?:markdown)?",
        "",
        brief.strip(),
        flags=re.IGNORECASE
    )

    brief = re.sub(
        r"```$",
        "",
        brief.strip()
    ).strip()

    # --------------------------------------------------------
    # Guarantee References section
    # --------------------------------------------------------

    if "References" not in brief:
        brief += "\n\nReferences\n\n"

        for i, source in enumerate(sources, 1):
            brief += (
                f"[{i}] {source['title']} — "
                f"{source['url']}\n"
            )

    # --------------------------------------------------------
    # Guarantee at least two short quotes if possible
    # --------------------------------------------------------

    quoted_matches = re.findall(
        r'"([^"]+)"',
        brief
    )

    valid_quotes = [
        q for q in quoted_matches
        if _count_words(q) <= 10
    ]

    if len(valid_quotes) < 2:

        quote_candidates = []

        for item in report.get("snippets", []):
            snippet = item.get("snippet", "").strip()

            if not snippet:
                continue

            # Take a short sentence
            sentences = re.split(
                r'(?<=[.!?])\s+',
                snippet
            )

            for sentence in sentences:
                words = re.findall(
                    r"\b[\w'-]+\b",
                    sentence
                )

                if 3 <= len(words) <= 10:
                    quote_candidates.append(
                        " ".join(words)
                    )

                if len(quote_candidates) >= 2:
                    break

            if len(quote_candidates) >= 2:
                break

        # Insert quotes before References
        if quote_candidates:
            references_index = brief.find("References")

            if references_index != -1:
                quote_section = "\n\n"

                for q in quote_candidates[:2]:
                    quote_section += (
                        f'As one source states, "{q}".\n'
                    )

                brief = (
                    brief[:references_index]
                    + quote_section
                    + brief[references_index:]
                )

    # --------------------------------------------------------
    # Save brief.md
    # --------------------------------------------------------

    brief_path = ARTIFACTS_DIR / "brief.md"

    with open(
        brief_path,
        "w",
        encoding="utf-8"
    ) as f:
        f.write(brief)

    # --------------------------------------------------------
    # Save updated report
    # --------------------------------------------------------

    report["brief"] = brief

    report["references"] = sources

    save_logs(
        report,
        str(ARTIFACTS_DIR / "report.json")
    )

    return {
        **report,
        "brief": brief,
        "references": sources,
    }


print("Cell 2 complete.")
print("Agent and all required tools are ready.")
````

````python
# ============================================================
# CELL 3 — PIPELINE + EVALUATION + RUN
# ============================================================

# ============================================================
# Q5 — PIPELINE
# ============================================================

def run_pipeline(topic: str) -> Dict[str, Any]:

    print("=" * 70)
    print("STARTING RESEARCH PIPELINE")
    print("=" * 70)
    print("Topic:", topic)
    print()

    # Q3
    print("1. Running research agent...")
    report = run_agent(
        topic,
        rounds=1
    )

    # Q4
    print("2. Synthesizing final brief...")
    report = synthesize_brief(
        report,
        min_paras=5,
        max_paras=8
    )

    # Save final report again
    save_logs(
        report,
        str(ARTIFACTS_DIR / "report.json")
    )

    print()
    print("Pipeline complete.")
    print("Saved:")
    print(" - artifacts/brief.md")
    print(" - artifacts/report.json")

    return report


# ============================================================
# EVALUATION
# ============================================================

def evaluate_outputs(
    brief_path: str = str(ARTIFACTS_DIR / "brief.md"),
    report_path: str = str(ARTIFACTS_DIR / "report.json")
) -> Dict[str, Any]:

    scorecard = {
        "has_references": False,
        "inline_citations": 0,
        "unique_domains": 0,
        "quotes": 0,
        "quotes_valid": 0,
        "report_exists": False,
        "brief_exists": False,
        "logs_exist": False,
        "passed": False,
    }

    # --------------------------------------------------------
    # Check brief
    # --------------------------------------------------------

    brief_file = pathlib.Path(brief_path)

    if brief_file.exists():
        scorecard["brief_exists"] = True

        brief = brief_file.read_text(
            encoding="utf-8"
        )

        scorecard["has_references"] = (
            "References" in brief
        )

        citations = re.findall(
            r"\[\d+\]",
            brief
        )

        scorecard["inline_citations"] = len(
            set(citations)
        )

        # Extract quoted text
        quotes = re.findall(
            r'"([^"]+)"',
            brief
        )

        scorecard["quotes"] = len(quotes)

        scorecard["quotes_valid"] = sum(
            1
            for q in quotes
            if _count_words(q) <= 10
        )

    # --------------------------------------------------------
    # Check report
    # --------------------------------------------------------

    report_file = pathlib.Path(report_path)

    if report_file.exists():

        scorecard["report_exists"] = True
        scorecard["logs_exist"] = True

        try:
            with open(
                report_file,
                "r",
                encoding="utf-8"
            ) as f:
                report = json.load(f)

            urls = []

            for source in report.get(
                "references",
                []
            ):
                url = source.get("url", "")
                if url:
                    urls.append(url)

            if not urls:
                urls = report.get(
                    "selected_urls",
                    []
                )

            domains = set()

            for url in urls:
                try:
                    domain = urlparse(url).netloc.lower()

                    if domain.startswith("www."):
                        domain = domain[4:]

                    if domain:
                        domains.add(domain)

                except Exception:
                    pass

            scorecard["unique_domains"] = len(domains)

        except Exception:
            pass

    # --------------------------------------------------------
    # Overall grading checks
    # --------------------------------------------------------

    scorecard["passed"] = all([
        scorecard["brief_exists"],
        scorecard["report_exists"],
        scorecard["logs_exist"],
        scorecard["has_references"],
        scorecard["inline_citations"] >= 1,
        scorecard["unique_domains"] >= 3,
        scorecard["quotes_valid"] >= 2,
    ])

    # Save scorecard
    save_logs(
        scorecard,
        str(ARTIFACTS_DIR / "scorecard.json")
    )

    return scorecard


# ============================================================
# RUN YOUR ASSIGNMENT
# ============================================================

TOPIC = "Artificial intelligence in education"

final_report = run_pipeline(TOPIC)

print()
print("=" * 70)
print("FINAL BRIEF")
print("=" * 70)
print()

print(
    final_report.get(
        "brief",
        "No brief generated."
    )
)

print()
print("=" * 70)
print("EVALUATION")
print("=" * 70)

scorecard = evaluate_outputs()

for key, value in scorecard.items():
    print(f"{key}: {value}")

print()
print("Files created:")
print("artifacts/brief.md")
print("artifacts/report.json")
print("artifacts/scorecard.json")
````
